# massimo, minimo, media dei votiù

```python
voti = "7;5.5;9;6;8"

massimo = 0.0
minimo = 10
somma = 0.0
media = 0.0

array_voti = []

for i in voti.split(";"):
    array_voti.append((float)(i))

for i in array_voti:
    if i > massimo:
        massimo = i
    if i < minimo:
        minimo = i
    somma += i


print("-" * 30)
print(f"massimo: {massimo:.2f}")
print(f"minimo:  {minimo:.2f}")
print(f"media:   {somma / len(array_voti) :.2f}")
```

# Massimo comun divisore con l'algoritmo di Euclide.

```python
a = 1071
b = 462

a = abs(a)
b = abs(b)

if a >= b:
    while b > 0:
        temp = a
        a = b
        b = temp % b

print(f"mdc: {a:.2f}")
```
